```python
#8. 소수 판별하기

num = int(input("정수를 입력하세요: "))

count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count = count + 1

if count == 2:
    print("소수입니다")
else:
    print("소수가 아닙니다")
```

```python
#9. 3~100 사이의 소수 개수 구하기

count = 0

for num in range(3, 101):
    divisor = 0

    for i in range(1, num + 1):
        if num % i == 0:
            divisor = divisor + 1

    if divisor == 2:
        count = count + 1

print("소수 개수:", count)
```

```python
#10. 별 출력하기

n=int(input("숫자를 입력하세요: "))

for i in range(1, n+1):
    print("*" * i)
```

```python
#11. X 모양 별 출력
n = 10

for i in range(n):
    for j in range(n):
        if j == i or j == n-1-i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
