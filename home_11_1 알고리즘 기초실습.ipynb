```python
#8. 소수 판별하기

num = int(input("정수를 입력하세요: "))

count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count = count + 1

if count == 2:
    print("소수입니다")
else:
    print("소수가 아닙니다")
```

```python
#9. 3~100 사이의 소수 개수 구하기

count = 0

for num in range(3, 101):
    divisor = 0

    for i in range(1, num + 1):
        if num % i == 0:
            divisor = divisor + 1

    if divisor == 2:
        count = count + 1

print("소수 개수:", count)
```

```python
#10. 별 출력하기

n=int(input("숫자를 입력하세요: "))

for i in range(1, n+1):
    print("*" * i)
```

```python
#11. X 모양 별 출력
n=int(input("숫자를 입력하세요: "))

for i in range(n):
    for j in range(n):
        if j == i or j == n-1-i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
#12. 가로 구구단 출력

#2단부터 9단까지를 가로 방향으로 출력하는 프로그램을 작성하시오.
#출력 포맷 답 부분은 2자리 숫자로 출력 ex) "2*1=2" -> "2*1= 2"

#2*1= 2 2*2= 4 2*3= 6 2*4= 8 2*5=10 2*6=12 2*7=14 2*8=16 2*9=18 
#3*1= 3 3*2= 6 3*3= 9 3*4=12 3*5=15 3*6=18 3*7=21 3*8=24 3*9=27 

for dan in range(2, 10):                                #단의 범위설정
    for num in range(1, 10):                            #곱셈 범위설정
        result = dan * num                              # ex)2*6
        print(f"{dan}*{num}={result:2}", end=" ")       # end="" 줄바꿈 안하고 옆으로 출력 위해 필요, result:2 결과값을 2자리로 표현하기 위해 필요
    print()                                             #줄바꿈 위해 필요
```

```python
# 13.세로 구구단 출력

for i in range(1,10):
    for dan in range(2,10):
        print(f"{dan} X {i}={dan*i:>2}",end="|")
    print("",end="\n")
```
